```python
import csv
import pandas as pd

# Read CSV file using pandas
df = pd.read_csv('/content/AI_DATASET_HACKATHON .csv')

# Get column names
column_names = df.columns.tolist()

# Create a list of lists with column names included
data_list = []
for index, row in df.iterrows():
    row_list = []
    for i, value in enumerate(row):
        row_list.append(f"{column_names[i]}: {value}")  # Combine column name and value
    data_list.append(row_list)

# Print the list of lists
for row_list in data_list:
    print(row_list)
    print("-" * 20)  # Separator between rows
```

```python
import csv
import openai
import pandas as pd
from pinecone import Pinecone

# Set API keys

PINECONE_API_KEY = userdata.get('Pinecone_API_Key')
PINECONE_ENVIRONMENT = 'us-east-1'



# Initialize Pinecone
pc = Pinecone(api_key=PINECONE_API_KEY, environment=PINECONE_ENVIRONMENT)
index = pc.Index("medical-paitents-dataset")

# OpenAI API Key
openai.api_key = userdata.get('openai_apikey')

# Function to generate embeddings using OpenAI

def get_embedding(text):
    response = openai.embeddings.create(
        input=text,
        model="text-embedding-3-small"  # model name
    )
    return response.data[0].embedding


# Upload data to embeddings and store in Pinecone
for i, row_list in enumerate(data_list):
    text = " ".join(row_list)  # Combine elements of the list into a single string
    embedding = get_embedding(text)
    row_id = f"row-{i}"  # Unique ID for each row
    index.upsert(vectors=[(row_id, embedding, {"text": text})])

    print(f"Stored: {row_id}")

print("Data successfully stored in Pinecone!")
```

```python
def search_similar_cases(query, index): # Pass the Pinecone index as an argument
    """Search Pinecone for similar medical cases"""
    query_embedding = get_embedding(query) # Generate embedding for the query
    results = index.query(vector=query_embedding, top_k=3, include_metadata=True) # Use the index for querying

    print("🔎 Similar Cases:")
    for res in results['matches']:
        print(f"- {res['metadata']['text']}") # Access text from metadata

# Example query
search_similar_cases("Severe headache and dizziness", index) # Pass the index to the function
```
